```python
### Um curso intesivo de Python

# Alguns postulados do python que devem ser seguido como:
# Um dos mais populares:
## Dever haver uma - preferivelmente, apenas um - forma evidente de fazer algo
### Quando o código é inscrito dessa forma evidente, é descrito como Pythonic

# Existem código escritos através de método Pythonic e não Pythonic

# OUtros princípios são mais estéticos 
## Outro postulado é: 
### É melhor algo belo do qu feio. Algoe expresso do que implícito. Algo simples do que complexo.

## Esses postulados orientarão nosso código
```

```python
## Funções
## CAda função é uma regra que receber zero ou mais entradas e retorna a saída correspondente

def double(x):
    "É colocado um docstring opaconal para descrever a função"
    return x*2

## Funções no python são funções de primeira classe, ou seja, podemos atribuí-las a variáveis e inserí-las nas funções como argumento
def apply_to_onde(f):
    "Chanma a função f usando 1 como argumento"
    return f(5)

# implmentação

my_doyble =  double
x = apply_to_onde(my_doyble)
x

## também é fácil criar funções anônimas chamadas lambdas

y =  apply_to_onde(lambda x: x+4)
y

# Também pode atribuir as funções lambdas a variáveis, embora o recomendado seja atribuir a funções
another_double =  lambda x: x*2 ## NÃO É PARA FAZER ISSO

def another_double(x):
    "Ao inves faça isso"
    return x * 2

# Os parametros das funções podem assumir argumentos padrão, que devem ser especificados se você quiser obter um valor diferente do padrão

def my_print(message = "Essa é minha mensagem padrão"):
    print(message)

my_print("Hello, World!!")
my_print()


# As vezes é útil especificar argumentos pelo nome
def full_name(fist_name = "What's his first name", last_name = "What's his last name"):
    return fist_name + " " + last_name

full_name("Valdemar", "Neto")
```
