```python
#Encontrando Conectores

#Data Dump da REde DataSciencester 
# Data dump é o um dict que conte´m os eu id e seu name 

users = [ # data dump em que cada user é representado por um dict com seu id e name
    {"id":0, "name": "Hero"},
    {"id":1, "name": "Dunn"},
    {"id":2, "name": "Sue"},
    {"id":3, "name": "Chi"},
    {"id":4, "name": "Thor"},
    {"id":5, "name": "Clive"},
    {"id":6, "name": "Hicks"},
    {"id":7, "name": "Devin"},
    {"id":8, "name": "Kate"},
    {"id":9, "name": "Klein"},
]

friendship_pairs = [(0,1), (0,2), (1,2), (1,3), (2,3), (3,4),
                    (4,5), (5,6), (5,7),(6,8),(7,8), (8,9)] # A tupla (0,1) representa que o usuario de id 0 "Hero" é amigo do usuario de id 1 "Dun"


## Utilizar essa abordagem de dois dumps não é eficiente quanto a busca entre amigos, pois necessário é iterar sobre todos os dados.
## O que será feito? Um novo dict que as chaves serão os id e os valores serão os ids dos seus amigos
## Porém, é preciso conferir todos os apres para criar o dict, mas só uma vez, então, depois, teremos pesquisas mais eficientes

# Novo dict inicializado vazio
friendship = {user["id"]: [] for user in users} 
friendship
```

```python
## Agora será criado o novo dict
for i, j in friendship_pairs: 
    friendship[i].append(j) # Adicionando j como amigo de i
    friendship[j].append(i) # Adicionando i como amigo de usuario j

friendship
```

```python
## Agora já temos os dict com o valor de todas as amizades
## Agora, perguntas mais complexas podem ser respondidas rapidamente
# Para isso vamos criar uma função para criar determinar o número total de conexões somanto o tamanho de todas as listas

def number_of_friend(user):
    "Quantos amigos tem o user"
    user_id = user["id"]
    friends_ids = friendship[user_id]
    return len(friends_ids)

number_of_friend(users[7])

total_connections = sum(number_of_friend(user) for user in users)

total_connections
```

```python
## Para encontrar o número médio de conexões basta pegar o toal de conexões e dividir pelo número de usuários

num_users = len(users)
print("Len(user):", num_users)
agv_connections = total_connections/num_users
print("Agv connections:", agv_connections)
```

```python
## Agora, podmos fazer outra busca como as pessoas mais conectadas. 
#Pode ser feito uma nova lista, com ordenação, no qual o chave será o id do user e o valor será quantos amigos tem na rede
# Ao fazer isso, temos uma abordagem chamdada de Centralidade de grau

num_friend_by_id = [(user["id"], number_of_friend(user)) for user in users]
print(num_friend_by_id)

num_friend_by_id.sort(
    key=lambda id_and_friends: id_and_friends[1],
    reverse=True
)

num_friend_by_id
```

```python

### Cientista de Dados que você talvez conheça
## Iremos implementar agora uma lista de foaf (friends of a friend)

def foaf_id(user):
    "Foaf = amigo do meu amigo"

    return [foaf_id 
            for friend_id in friendship[user["id"]]
            for foaf_id in friendship[friend_id]
            ]

print("friendship 0:", friendship[0])
print("Id 0:",foaf_id(users[0]))
print("friendship 1:", friendship[1])
print("Id 1:",foaf_id(users[1]))
print("friendship 2:", friendship[2])
print("Id 2:",foaf_id(users[2]))
print("friendship 3:", friendship[3])
print("Id 3:",foaf_id(users[3]))
```

```python
### Agora, que já conhecemos todos os dados de foaf podemos fazer um novo dict eliminando os amigos que já conhecem
## Utilizando a biblioteca de Counter 

from collections import Counter

def friends_of_friends(users):
    user_id = users["id"]
    return Counter(
        foaf_id
        for friend_id in friendship[user_id] #Para cada amigo meu
        for foaf_id in friendship[friend_id] #encontre os amigos deles
        if foaf_id != user_id  # que não sejam eles
        and foaf_id not in friendship[user_id] # e não sejam meus amigos.
    )

print(friends_of_friends(users[3])) ## Irá retornar que Chi (3) que possui dois amigos em comum com id 0 e um amigo em comum com o id 5
```

```python
### Agora que já temos uma boa abordagem sobre os pares de amigos, podemos implementar outros dumps
##  Agora querermos encontrar usuarios com interesses similares, temos um novo dump interests
## CIENTISTA QUE TALVEZ VOCE CONHEÇA

interests = [
    (0,"Hadoop"),(0,"Big Data"),(0,"HBase"),(0,"Java"),
    (0,"Spark"),(0,"Storm"),(0,"Cassandra"),
    (1,"NoSQL"),(1,"MongoDB"),(1,"Cassandra"),(1,"HBase"),
    (1, "Postgres"),(2,"Python"),(2,"scikit-learn"),(2,"scipy"),
    (2,"numpy"),(2,"statsmodels"),(2,"pandas"),(3,"R"),(3,"Python"),
    (3,"statistics"),(3,"regression"),(3,"probability"),
    (4,"machine learning"),(4,"regression"),(4,"decision trees"),
    (4,"libsvm"),(5,"Python"),(5,"R"),(5,"Java"),(5,"C++"),
    (5,"Haskell"),(5,"programming languages"),(6,"statistics"),
    (6,"probability"),(6,"mathematics"),(6,"theory"),
    (7,"machine learning"),(7,"scikit-learn"),(7,"Mahout"),
    (7,"neural networks"),(8,"neural networks"),(8,"deep learning"),
    (8,"Big Data"),(8,"artificial intelligence"),(9,"Hadoop"),
    (9,"Java"),(9,"MapReduce"),(9,"Big Data")
]

## Com isso, apesar de não serem amigos, podemos encontrar usuarios que tem os mesmos interesses como é o caso do user_id 0 e user_id9, ambos 
## se interessam por Java e BigData
## Nesse caso, podemos fazer um função que realize essa busca para verificar os interesses similares

def data_scientists_who_like(target_interest):
    "Encontra usuarios de mesmo interesses"
    return [user_id
            for user_id, user_interest in interests # dentro de interest, procure usuarios de mesmo id e o mesmo interesses
            if user_interest == target_interest] # que tenham os mesmos interesses


# Indice de interesses é utilizado para examinar a lista de interesses inteira

from collections import defaultdict

# as chaves são os interesses, os valores são listas de user_ids com o interesse em questão 
user_id_by_interest = defaultdict(list)

#Construção de indices de interesses
for user_id, interest in interests:
    user_id_by_interest[interest].append(user_id)

# Construção de outro indice, de usurarios para interesse. AS chaves são os user_id e o valoes são listas de interesses do user_id emq questão
interest_by_user_id = defaultdict(list)
for user_id, interest in interests:
    interest_by_user_id[user_id].append(interest)

print("Indice de interesse de usuarios", user_id_by_interest)
print("Indice de usuarios", interest_by_user_id)


#Ficou fácil determinar quem tem mais interesses em comum com um usuário específico
# Passo a passo: 
## Interar entre os usuários interessados
## Cada interesse, itere sobre os outros usuários com o mesmo interesse
## Depois contar quantas vezes cada usuário aparece

def most_comom_interest(user):
    return Counter(
        interested_user_id
        for interest in interest_by_user_id[user["id"]]
        for interested_user_id in user_id_by_interest[interest]
        if interested_user_id != user["id"]
    )

most_comom_interest(users[1])
```

```python
### Agora, trabalhando com ideias de Salários e Experiencias
### Tópico é sensível e fornecem um conjunto de dados anonimos

salaries_and_tenures = [
    (83000, 8.7), (88000, 8.1),
    (48000, 0.7), (76000, 6),
    (69000, 6.5), (76000, 7.5),
    (60000, 2.5), (83000, 10),
    (48000, 1.9), (63000, 4.2)
]

# ## Plotando esses dados

# # # IMPRESSÃO DO GRÁFICO:
salaries_and_tenures.sort(key=lambda salaries_and_tenures: salaries_and_tenures[1], # ordena a lista decrescentemente pelo tenure
reverse=True)

# # # Descompacte as tuplas em duas listas
# salaries, tenures = zip(*salaries_and_tenures)

# import matplotlib.pyplot as plt
# plt.plot(tenures, salaries, 'o')  # Adicionando marcadores " marker='o' " para melhor visualização
# plt.xlabel('Experiência (Anos)') # Rótulo eixo x
# plt.ylabel('Salário ($/Ano)') # Rótulo eixo y
# plt.title('Salário vs. Experiência') # Título do gráfico
# plt.grid()
# plt.show() # Exibir o gráfico

#### Analisando a média salarial por anos de experiencia
salary_by_tenure = defaultdict(list)

for salary, tenure in salaries_and_tenures:
    salary_by_tenure[tenure].append(salary)

# as chaves são anos, cada valor é o salário médio associado ao número de anos de experiência
avg_salary_by_tenure = {
    tenure: sum(salaries) / len(salaries) for tenure, salaries in salary_by_tenure.items()
}

avg_salary_by_tenure

sorted(avg_salary_by_tenure.items())
```

```python
## Esse estilo não parece muito útil, já que todos os usuários não tem os mesmos anos de experiência
## Fazendo buckets de experiencia

def tenure_bucket(tenure):
    if tenure < 2:
        return "less than two"
    elif tenure < 5:
        return "between two and five"
    else: return "more than five"

## Agrupando cada salário em um bucket especifico

salary_by_tenure_bucket = defaultdict(list)
for salary, tenure in salaries_and_tenures:
    bucket = tenure_bucket(tenure)
    salary_by_tenure_bucket[bucket].append(salary)

# Computando as médias salarial de cada grupo 
avg_salary_by_bucket = {
    tenure_bucket: sum(salaries) / len(salaries)
    for tenure_bucket, salaries in salary_by_tenure_bucket.items()
}

sorted(avg_salary_by_bucket.items())

## Perceba-se que os cientistas de dados com mais cinco ano sde experiencia ganham 65% mais do que os colegas com pouca ou nenhuma experiência
```
